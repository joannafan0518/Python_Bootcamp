## Pandas

```python
import pandas as pd
```

#### 1. Create a pandas DataFrame named `df` using the following table of names, ages, and scores.

| Name  | Age | Score |
|-------|-----|-------|
| Anna  | 23  | 88    |
| Ben   | 35  | 92    |
| Cara  | 29  | 79    |
| David | 41  | 85    |

```python
# Your code goes here
df = pd.DataFrame({
    'Name': ['Anna', 'Ben', 'Cara', 'David'],
    'Age': [23, 35, 29, 41],
    'Score': [88, 92, 79, 85]
})

df
```

#### 2. Use pandas methods on `df` to find the dataset’s dimensions (number of rows and columns) and calculate the average score.

```python
# Your code goes here
print(df.shape)
```

```python
print(df['Score'].mean())
```

#### 3. Filter the DataFrame `df` to keep only rows where the score is greater than or equal to 85 **and** the age is greater than 30.


```python
# Your code goes here
df.loc[(df['Score'] >= 85) & (df['Age'] > 30)]
```

#### 4. Create a new column `Score_decimal` by dividing the `Score` column by 100.

```python
# Your code goes here
df['Score_decimal'] = df['Score'] / 100

df
```

## Matplotlib

```python
import matplotlib.pyplot as plt
```

#### 1. **Bar Plot.** Create a bar chart showing each person’s Name on the x-axis and their Score on the y-axis.

```python
# Your code goes here
fig, ax = plt.subplots()

ax.bar(df['Name'], df['Score'])
ax.set_title('Score by Name')
ax.set_xlabel('Name')
ax.set_ylabel('Score')

plt.show()
```

#### 2. **Scatter Plot.** Create a scatter plot for df showing the relationship between `Age ` and `Score`.

```python
# Your code goes here
fig, ax = plt.subplots()

ax.scatter(df['Age'], df['Score'])
ax.set_title('Age vs Score')
ax.set_xlabel('Age')
ax.set_ylabel('Score')

plt.show()
```

#### 3. **Histogram**. Create a histogram to show the distribution of the `Score` column.

```python
# Your code goes here
fig, ax = plt.subplots()

ax.hist(df['Score'])
ax.set_title('Score Distribution')
ax.set_xlabel('Score')
ax.set_ylabel('Frequency')

plt.show()
```
